```python
import pandas as pd
import sqlite3
```

```python
data = {
   "store": ["store A", "store B", "store C", "store D"],
   "sales": [50000, 20000, 70000, 18000],
   "inventory": [10000, 12000, 15000, 9000],
   "returns": [3000, 4000, 1000, 5000]
}

df = pd.DataFrame(data)

df
```

```python
df["return_rate"] = (
    df["returns"] / df["sales"]
)
df
```

```python
def generate_recomendation(row):

    if row["sales"] < 25000:
       return "increase marketing efforts"

    elif row["return_rate"] > 0.15:
       return "investigate product quality"

    else:
        return "maintain current startegy"
```

```python
df["recommendation"] = df.apply(
    generate_recomendation,
    axis=1
)

df
```

```python
conn = sqlite3.connect(
    "consulting_analysis.db"
)

df.to_sql(
    "store_analysis",
    conn,
    if_exists="replace",
    index=False
)

print("saved to SQL database")
```

```python
df.to_excel(
    "recommendation.xlsx",
    index=False
)

print("excel report created")
```

```python
print("==============================")
print("Consulting Workflow Automation")
print("==============================")

print("\nStores analyzed:", len(df))
print("\nRecommendations:")
for _, row in df.iterrows():
    print(
        f"{row['store']} -> {row['recommendation']}"
    )
```
